```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries Loaded Successfully ✅")
```

```python
transaction = pd.read_csv("../data/raw/train_transaction.csv")
identity = pd.read_csv("../data/raw/train_identity.csv")

print("Transaction Shape :", transaction.shape)
print("Identity Shape    :", identity.shape)
```

```python
train = transaction.merge(
    identity,
    on="TransactionID",
    how="left"
)

print("Merged Dataset Shape :", train.shape)
```

```python
fraud_counts = train["isFraud"].value_counts()

fraud_percent = (
    train["isFraud"]
    .value_counts(normalize=True)*100
)

display(pd.DataFrame({
    "Count": fraud_counts,
    "Percentage": fraud_percent.round(3)
}))
```

```python
plt.figure(figsize=(6,6))

plt.pie(
    fraud_counts,
    labels=["Legitimate","Fraud"],
    autopct="%1.2f%%",
    startangle=90
)

plt.title("Fraud Distribution")

plt.show()
```

```python
plt.figure(figsize=(12,5))

sns.histplot(
    train["TransactionAmt"],
    bins=100,
    kde=True
)

plt.title("Transaction Amount Distribution")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    x="isFraud",
    y="TransactionAmt",
    data=train
)

plt.yscale("log")

plt.title("Fraud vs Legitimate Transactions")

plt.show()
```

```python
plt.figure(figsize=(12,5))

sns.histplot(
    train["TransactionAmt"],
    bins=100,
    kde=True
)

plt.title("Transaction Amount Distribution")
plt.xlabel("Transaction Amount")
plt.ylabel("Count")

plt.show()
```

## Observation

Most transactions are of relatively low value, while a few transactions have very high amounts.

## Business Insight

Extremely high-value transactions are uncommon and may warrant additional verification, especially when combined with other suspicious indicators.

## Feature Engineering Opportunity

- High-value transaction flag
- Transaction amount percentile
- Amount deviation from average

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    x="isFraud",
    y="TransactionAmt",
    data=train
)

plt.yscale("log")

plt.title("Fraud vs Legitimate Transaction Amount")
plt.xlabel("Fraud")
plt.ylabel("Transaction Amount (Log Scale)")

plt.show()
```

## Observation

Fraudulent transactions show a wider spread of transaction amounts compared to legitimate transactions.

## Business Insight

Transaction amount alone is not enough to identify fraud, but it becomes a powerful feature when combined with customer behavior and transaction context.

## Feature Engineering Opportunity

- Spending spike score
- High amount flag
- Amount percentile

```python
plt.figure(figsize=(16,8))

sns.heatmap(
    train.isnull(),
    cbar=False,
    cmap="viridis"
)

plt.title("Missing Values Heatmap")

plt.show()
```

## Observation

Several identity-related features contain significant missing values.

## Business Insight

Real-world banking datasets often have incomplete information. The model must remain effective even when some identity attributes are unavailable.

## Feature Engineering Opportunity

- Missing value indicators
- Smart imputation
- Feature selection

```python
plt.figure(figsize=(12,5))

sns.histplot(
    train["TransactionDT"],
    bins=100,
    kde=True
)

plt.title("Transaction Time Distribution")
plt.xlabel("TransactionDT")
plt.ylabel("Count")

plt.show()
```

## Observation

Transaction times are not uniformly distributed, indicating periods of higher transaction activity.

## Business Insight

Fraudsters often exploit periods of high transaction volume to hide fraudulent activities.

## Feature Engineering Opportunity

- Hour of transaction
- Time bucket (Morning/Afternoon/Evening/Night)
- Peak hour indicator

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=train,
    x="ProductCD",
    order=train["ProductCD"].value_counts().index
)

plt.title("Transactions by Product Type")

plt.show()
```

```python
fraud_product = pd.crosstab(
    train["ProductCD"],
    train["isFraud"],
    normalize="index"
)*100

fraud_product
```

```python
fraud_product.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Fraud Percentage by Product Type")
plt.ylabel("Percentage")

plt.show()
```

## Observation

Certain product categories have noticeably higher fraud rates than others.

## Business Insight

Product type is an important contextual feature and should be considered during fraud risk evaluation.

## Feature Engineering Opportunity

- Product risk score
- High-risk product flag

```python
plt.figure(figsize=(12,5))

train["card4"].value_counts().plot(kind="bar")

plt.title("Card Network Distribution")

plt.show()
```

```python
card_fraud = pd.crosstab(
    train["card4"],
    train["isFraud"],
    normalize="index"
)*100

card_fraud
```

```python
card_fraud.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Fraud Rate by Card Network")

plt.ylabel("Percentage")

plt.show()
```

## Observation

Fraud rates differ across card networks.

## Business Insight

Card network information can contribute to fraud prediction but should be combined with other transaction features.

## Feature Engineering Opportunity

- Card network encoding
- Card network fraud score
